```python
import findspark
findspark.init()

from pyspark.sql import SparkSession
from pyspark.sql.functions import col, avg, desc
import os

# Start Spark session first
spark = SparkSession.builder.appName("StudentPerformance").getOrCreate()

# Resolve path to CSV in same folder as notebook
path = "file:///" + os.path.abspath("student_performance.csv").replace("\\", "/")
print("Resolved path:", path)

# Load dataset
data = spark.read.csv(path, header=True, inferSchema=True)

# Show data
data.show(5)
data.printSchema()
```

```python
# Average marks per section
data.groupBy("Section").agg(avg("Marks")).show()

# Top 5 performers
data.orderBy(col("Marks").desc()).show(5)
```

```python
rows = data.select("StudyHours", "Marks").collect()

x = [row["StudyHours"] for row in rows]
y = [row["Marks"] for row in rows]

import matplotlib.pyplot as plt
plt.scatter(x, y)
plt.xlabel("Study Hours")
plt.ylabel("Marks")
plt.title("Study Hours vs Marks")
plt.show()
```
